# General Simple Forward Example

This notebook uses the symbolic API directly and samples parameters from a simplex region.
        

```python
from pathlib import Path
import sys

import numpy as np

ROOT = next(
    path for path in [Path.cwd(), *Path.cwd().parents]
    if (path / 'nlpoptnet').is_dir() and (path / 'notebooks').is_dir()
)
SRC = ROOT / 'nlpoptnet' / 'src'
if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))

from nlpoptnet import NLPOptNet
```

```python
CONFIG = {
    'epochs': 3,
    'batch_size': 8,
    'learning_rate': 1e-3,
    'train_frac': 0.8,
    'hidden_size': 32,
    'hidden_layers': 2,
    'seed': 42,
    'dtype': 'float64',
    'print_every': 1,
    'verbose': True,
}
```

## Define a symbolic general problem
        

```python
model = NLPOptNet(config=CONFIG, name='notebook_general_simple')
x = model.add_parameter(['x1', 'x2'])
y = model.add_variable(['y1', 'y2'])

model.objective(0.5 * (y.y1 ** 2 + y.y2 ** 2))
model.constraints.equality.add(
    y.y1 + y.y2 == x.x1 + 0.5 * x.x2,
)
model.constraints.inequality.add(
    y.y1 ** 2 + y.y2 ** 2 <= 1.5,
)
model.constraints.box.add(
    y.y1 >= -1.0,
    y.y2 >= -1.0,
    y.y1 <= 1.0,
    y.y2 <= 1.0,
)
model.simplex(
    x.x1 >= 0.0,
    x.x2 >= 0.0,
    x.x1 + x.x2 <= 1.0,
    num_samples=32,
)
model.build()
result = model.optimize()
run_dir = Path(result['output_dir'])
run_dir
        
```

```python
reloaded = NLPOptNet().load(run_dir / 'metadata.json')
reloaded.predict([0.25, 0.5])
        
```
